```python
from datetime import date, timedelta

habitslist = []

print("Welcome to Habit Tracker App")

while True:
    print("\n==== MENU ====")
    print("1. Add Habit")
    print("2. View Habits")
    print("3. Mark Habit as Done Today")
    print("4. View Streaks")
    print("5. Delete Habit")
    print("6. Exit")

    choice = input("Please Enter Your Choice: ")

    # 1.
    if choice == "1":
        print("---------------- Add Your Habit ----------------")
        name = input("Enter habit name (e.g. Exercise, Read, Meditate): ").strip()

        if name == "":
            print("Habit name cannot be empty.")
            continue

        # check for duplicates (ignoring upper/lower case)
        duplicate = False
        for habit in habitslist:
            if habit["name"].lower() == name.lower():
                duplicate = True
        if duplicate:
            print("This habit already exists.")
            continue

        habit = {
            "name": name,
            "dates": [],          # list of dates (YYYY-MM-DD) on which habit was done
        }

        habitslist.append(habit)
        print("\nHabit added successfully")

    # 2.
    elif choice == "2":
        if not habitslist:
            print("No Habits Added.")
        else:
            today = str(date.today())
            print("\n==== All Habits ====")
            for count, habit in enumerate(habitslist, start=1):
                if today in habit["dates"]:
                    status = "Done today"
                else:
                    status = "Not done today"
                print(
                    f"Habit {count} -> {habit['name']}, "
                    f"{status}, "
                    f"Total days: {len(habit['dates'])}"
                )

    # 3.
    elif choice == "3":
        if not habitslist:
            print("No Habits Added.")
            continue

        for count, habit in enumerate(habitslist, start=1):
            print(f"{count}. {habit['name']}")

        try:
            number = int(input("Enter habit number to mark as done: "))
        except ValueError:
            print("Invalid number. Please enter a number.")
            continue

        if number < 1 or number > len(habitslist):
            print("Habit number not found.")
            continue

        habit = habitslist[number - 1]
        today = str(date.today())

        if today in habit["dates"]:
            print("Already marked as done today.")
        else:
            habit["dates"].append(today)
            print(f"\nGreat job! '{habit['name']}' marked as done for today.")

    # 4.
    elif choice == "4":
        if not habitslist:
            print("No Habits Added.")
        else:
            print("\n==== Streaks ====")
            for habit in habitslist:
                done_days = set(habit["dates"])

                # current streak: count back from today (or yesterday if today not done yet)
                day = date.today()
                if str(day) not in done_days:
                    day = day - timedelta(days=1)
                current = 0
                while str(day) in done_days:
                    current += 1
                    day = day - timedelta(days=1)

                # longest streak
                longest = 0
                run = 0
                previous = None
                for d in sorted(done_days):
                    d = date.fromisoformat(d)
                    if previous is not None and d - previous == timedelta(days=1):
                        run += 1
                    else:
                        run = 1
                    if run > longest:
                        longest = run
                    previous = d

                print(
                    f"{habit['name']} -> Current streak: {current}, "
                    f"Longest streak: {longest}"
                )

    # 5.
    elif choice == "5":
        if not habitslist:
            print("No Habits Added.")
            continue

        for count, habit in enumerate(habitslist, start=1):
            print(f"{count}. {habit['name']}")

        try:
            number = int(input("Enter habit number to delete: "))
        except ValueError:
            print("Invalid number. Please enter a number.")
            continue

        if number < 1 or number > len(habitslist):
            print("Habit number not found.")
        else:
            removed = habitslist.pop(number - 1)
            print(f"\n'{removed['name']}' deleted successfully")

    # 6.
    elif choice == "6":
        print("Dhanyavad🙏")
        break

    else:
        print("INVALID CHOICE. TRY AGAIN")
```
